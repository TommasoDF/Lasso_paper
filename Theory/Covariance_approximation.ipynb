# Computing Cov(exp(x), x) by Taylor expansion and Stein's lemma

For $x \sim \mathcal{N}(0, \sigma^2_x)$, Stein's Lemma allows to compute the $Cov(g(x), x) = \sigma^2_x E(g'(x))$.

- 1) Compute the covariance for $g(x) = e^x$ by Stein's Lemma and compare it to the one obtained by first using the inifinite-order taylor series of $e^x$ and then computing the covariance.

- a) $g'(x) = e^x$, then $E(e^x) = e^{\sigma^2_x / 2}$, since $e^x$ is log-normal and $Cov(e^x,x) = \sigma^2_x e ^ {\sigma^2_x / 2}$
- b) The taylor series for $e^x$ around $x = 0$ is $\sum^{\infty}_{n= 0} \frac{x^n}{n!}$. Using $Cov(g(x), x) = E(g(x) x) - E(g(x))E(x)$, the fact that $E(x) = 0$, and the linearity of the expected value, we have 
$$ Cov(e^x, x) = E(e^x x) = \sum^{\infty}_{n= 0} \frac{E(x^{n+1})}{n!} $$

Now for a $0$ mean normal distribution 
$$ E(x^p) = \begin{cases}
0 \quad \textrm{if p is odd} \\
\sigma^p_x (p - 1) !! \quad \textrm{if p is even} 
\end{cases} $$

where $p!!$ denotes double factorial, that is, the product of all numbers from n to 1 that have the same parity as n.
Now let $n = 2k+1$, then 

$$E(x^{n+1})= E(x^{2k+2}) = (2k + 1)!! \sigma^{2k+2}_x$$
this gives 
$$Cov(e^x, x) =  \sum^{\infty}_{k= 0} \frac{(2k + 1)!! \sigma^{2k+2}_x}{(2k+1)!}$$

we then use the property $$(2k +1) !! = \frac{(2k +1)!}{2^k k!}$$,
to rewrite
$$Cov(e^x, x) =  \sum^{\infty}_{k= 0} \frac{\sigma^{2k+2}_x}{2^k k!} = \sigma^2_x \sum^{\infty}_{k= 0} \frac{\sigma^{2k}_x}{2^k k!} = \sigma^2_x \sum^{\infty}_{k= 0} \frac{(\sigma^{2}_x / 2)^k}{k!} = \sigma^2_x e^{\sigma^2_x / 2}$$
